### Ranking Attendance Percentage

- `rank()` assigns ranks to attendance values.
- `ascending=False` gives Rank 1 to the highest attendance.
- `method='first'` assigns different ranks to duplicate values based on their original order.
- `astype(int)` converts floating-point ranks into integers.
- `record['rank']` creates a new column to store the calculated ranks.

```python
record['rank'] = record['attendance_percentage'].rank(ascending=False, method='first').astype(int)
```

### Sorting DataFrame

- `sort_values()` sorts DataFrame rows.
- `by='rank'` specifies the column used for sorting.
- By default, sorting is in ascending order.
- `record =` stores the sorted DataFrame back into record.

```python
record = record.sort_values(by='rank')
```

### Displaying Top 10 Records

- `record[[...]]` selects specific columns.
- Double square brackets are used to select multiple columns.
- `head(10)` displays the first 10 rows.
- Since the DataFrame is already sorted by rank, it displays the top 10 ranked records.

```python
record[['attendance_percentage','rank']].head(10)
```
